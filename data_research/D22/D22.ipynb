```python
#주요 미국 경제지표
FRED_SERIES = {
    # 물가
    "cpi": "CPIAUCSL",
    "core_cpi": "CPILFESL",
    "pce": "PCEPI",
    "core_pce": "PCEPILFE",

    # 고용
    "unemployment_rate": "UNRATE",
    "nonfarm_payroll": "PAYEMS",

    # 경기 / 성장
    "real_gdp": "GDPC1",
    "industrial_production": "INDPRO",
    "retail_sales": "RSAFS",

    # 금리
    "fed_funds_rate": "DFF",
    "treasury_2y": "DGS2",
    "treasury_10y": "DGS10",
    "treasury_10y_2y_spread": "T10Y2Y",

    # 소비 심리
    "consumer_sentiment": "UMCSENT",
}
```

```python
import os
import requests


FRED_API_KEY = os.getenv('FRED_API_KEY')


def fetch_fred_vintage_dates(series_id):
    url = "https://api.stlouisfed.org/fred/series/vintagedates"

    params = {
        "series_id": series_id,
        "api_key": FRED_API_KEY,
        "file_type": "json",
    }

    response = requests.get(
        url,
        params=params,
        timeout=30,
    )

    response.raise_for_status()

    data = response.json()

    return data["vintage_dates"]
```

```python
# 해당 시리즈의 데이터 값이 수정되었거나, 새로운 관측값이 처음 추가된 날짜를 vintage_date라고 함

vintage_dates = fetch_fred_vintage_dates("GDPC1")

vintage_dates[-10:]
```

```python
def fetch_fred_vintage(
    series_id,
    vintage_date,
    start_date=None,
    end_date=None,
):
    url = "https://api.stlouisfed.org/fred/series/observations"

    params = {
        "series_id": series_id,
        "api_key": FRED_API_KEY,
        "file_type": "json",
        "vintage_dates": vintage_date,
    }

    if start_date is not None:
        params["observation_start"] = start_date

    if end_date is not None:
        params["observation_end"] = end_date

    response = requests.get(
        url,
        params=params,
        timeout=30,
    )

    response.raise_for_status()

    return response.json()
```

```python
# 빈티지 날짜를 기준으로 그 날짜 시점에 알려진 GDPC1의 전체 값을 불러와서 받는 것
data = fetch_fred_vintage(
    series_id="GDPC1",
    vintage_date="2025-08-01",
)
display(data)

#만약 최신값 하나만 받고 싶다면

latest = data["observations"][-1]
print("빈티지 날짜:", latest["realtime_start"])
print("관측 날짜:", latest["date"])
print("값:", latest["value"])
```
