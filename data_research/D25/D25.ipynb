````python
import requests
import pandas as pd
from io import StringIO


# ============================================================
# 1. 조회 조건
# ============================================================

START_YEAR = 2000
END_YEAR = 2025

# OECD National Accounts at a Glance - Chapter 1: GDP
BASE_URL = (
    "https://sdmx.oecd.org/public/rest/data/"
    "OECD.SDD.NAD,DSD_NAAG@DF_NAAG_I,1.0/"
)

# SDMX key
#
# A                  : Annual
# KOR+USA            : Korea + United States
# B1GQ_R_GR          : Real gross domestic product growth rate
# ..                 : 나머지 차원은 해당 지표에 대응되는 값 사용
#
SDMX_KEY = "A.KOR+USA.B1GQ_R_GR.."

url = (
    f"{BASE_URL}{SDMX_KEY}"
    f"?startPeriod={START_YEAR}"
    f"&endPeriod={END_YEAR}"
    f"&dimensionAtObservation=AllDimensions"
    f"&format=csvfilewithlabels"
)

print("요청 URL:")
print(url)
````

````python
response = requests.get(
    url,
    headers={
        "User-Agent": "Mozilla/5.0"
    },
    timeout=60
)

response.raise_for_status()

df = pd.read_csv(StringIO(response.text))

print("데이터 크기:", df.shape)
display(df.head())
````

````python
print("전체 칼럼 수:", len(df.columns))

for i, col in enumerate(df.columns, start=1):
    print(f"{i:2d}. {col}")
````

````python
important_columns = [
    "FREQ",
    "REF_AREA",
    "MEASURE",
    "UNIT_MEASURE",
    "PRICE_BASE",
    "TRANSACTION",
    "TRANSFORMATION",
    "TIME_PERIOD",
    "OBS_VALUE",
    "OBS_STATUS",
]

important_columns = [
    col for col in important_columns
    if col in df.columns
]

display(df[important_columns].head(20))
````

````python
check_columns = [
    "FREQ",
    "MEASURE",
    "UNIT_MEASURE",
    "PRICE_BASE",
    "TRANSACTION",
    "TRANSFORMATION",
]

for col in check_columns:
    if col in df.columns:
        values = df[col].dropna().unique()

        print(f"{col}: {values}")

        if len(values) != 1:
            print(
                f"  ⚠️ {col}에 여러 값이 존재합니다. "
                "한국과 미국을 직접 비교하기 전에 확인이 필요합니다."
            )
````

````python
same_condition_columns = [
    "FREQ",
    "MEASURE",
    "UNIT_MEASURE",
    "PRICE_BASE",
    "TRANSACTION",
    "TRANSFORMATION",
]

for col in same_condition_columns:
    if col not in df.columns:
        continue

    unique_values = df[col].dropna().unique()

    if len(unique_values) != 1:
        raise ValueError(
            f"{col} 기준이 하나로 통일되어 있지 않습니다: "
            f"{unique_values}"
        )

print("✅ 한국과 미국 데이터의 주요 단위·기준이 동일합니다.")
````

````python
countries = set(df["REF_AREA"].dropna().unique())

expected_countries = {"KOR", "USA"}

if countries != expected_countries:
    raise ValueError(
        f"예상 국가와 실제 국가가 다릅니다.\n"
        f"예상: {expected_countries}\n"
        f"실제: {countries}"
    )

print("✅ KOR, USA 데이터가 모두 존재합니다.")
````

````python
label_columns = [
    "Reference area",
    "Measure",
    "Unit of measure",
    "Frequency of observation",
    "Price base",
    "Time period",
    "Observation value",
]

for col in label_columns:
    if col in df.columns:
        print(f"\n[{col}]")
        print(df[col].dropna().unique()[:10])
````

````python
gdp = df[
    [
        "REF_AREA",
        "TIME_PERIOD",
        "OBS_VALUE",
        "MEASURE",
        "UNIT_MEASURE",
        "PRICE_BASE",
        "FREQ",
    ]
].copy()

gdp["OBS_VALUE"] = pd.to_numeric(
    gdp["OBS_VALUE"],
    errors="coerce"
)

gdp["TIME_PERIOD"] = pd.to_numeric(
    gdp["TIME_PERIOD"],
    errors="coerce"
)

gdp = (
    gdp
    .dropna(subset=["TIME_PERIOD", "OBS_VALUE"])
    .sort_values(["TIME_PERIOD", "REF_AREA"])
    .reset_index(drop=True)
)

display(gdp.head(20))
````

````python
comparison = (
    gdp
    .pivot(
        index="TIME_PERIOD",
        columns="REF_AREA",
        values="OBS_VALUE"
    )
    .rename(
        columns={
            "KOR": "Korea",
            "USA": "United States",
        }
    )
    .reset_index()
)

comparison.columns.name = None

display(comparison)
````

````python
import requests
import pandas as pd
from io import StringIO


def get_oecd_kor_usa_real_gdp_growth(
    start_year=2000,
    end_year=2025,
):
    base_url = (
        "https://sdmx.oecd.org/public/rest/data/"
        "OECD.SDD.NAD,DSD_NAAG@DF_NAAG_I,1.0/"
    )

    # Annual / KOR+USA / Real GDP growth rate
    sdmx_key = "A.KOR+USA.B1GQ_R_GR.."

    url = (
        f"{base_url}{sdmx_key}"
        f"?startPeriod={start_year}"
        f"&endPeriod={end_year}"
        f"&dimensionAtObservation=AllDimensions"
        f"&format=csvfilewithlabels"
    )

    response = requests.get(
        url,
        headers={"User-Agent": "Mozilla/5.0"},
        timeout=60,
    )
    response.raise_for_status()

    df = pd.read_csv(StringIO(response.text))

    if df.empty:
        raise ValueError("OECD API에서 데이터가 반환되지 않았습니다.")

    # --------------------------------------------------------
    # 국가 확인
    # --------------------------------------------------------
    countries = set(df["REF_AREA"].dropna().unique())

    expected_countries = {"KOR", "USA"}

    if countries != expected_countries:
        raise ValueError(
            f"국가가 예상과 다릅니다: {countries}"
        )

    # --------------------------------------------------------
    # 비교 기준 동일 여부 확인
    # --------------------------------------------------------
    comparison_dimensions = [
        "FREQ",
        "MEASURE",
        "UNIT_MEASURE",
        "PRICE_BASE",
        "TRANSACTION",
        "TRANSFORMATION",
    ]

    metadata = {}

    for col in comparison_dimensions:

        if col not in df.columns:
            continue

        values = df[col].dropna().unique().tolist()

        metadata[col] = values

        if len(values) != 1:
            raise ValueError(
                f"{col} 값이 하나로 통일되어 있지 않습니다: "
                f"{values}"
            )

    # --------------------------------------------------------
    # 필요한 칼럼
    # --------------------------------------------------------
    result = df[
        [
            "REF_AREA",
            "TIME_PERIOD",
            "OBS_VALUE",
            "FREQ",
            "MEASURE",
            "UNIT_MEASURE",
            "PRICE_BASE",
        ]
    ].copy()

    result["TIME_PERIOD"] = pd.to_numeric(
        result["TIME_PERIOD"],
        errors="coerce",
    )

    result["OBS_VALUE"] = pd.to_numeric(
        result["OBS_VALUE"],
        errors="coerce",
    )

    result = (
        result
        .dropna(subset=["TIME_PERIOD", "OBS_VALUE"])
        .sort_values(["TIME_PERIOD", "REF_AREA"])
        .reset_index(drop=True)
    )

    # --------------------------------------------------------
    # 비교용 wide table
    # --------------------------------------------------------
    comparison = (
        result
        .pivot(
            index="TIME_PERIOD",
            columns="REF_AREA",
            values="OBS_VALUE",
        )
        .rename(
            columns={
                "KOR": "Korea",
                "USA": "United States",
            }
        )
        .reset_index()
    )

    comparison.columns.name = None

    return df, result, comparison, metadata, url
````

````python
raw_df, gdp_df, comparison_df, metadata, url = (
    get_oecd_kor_usa_real_gdp_growth(
        start_year=2000,
        end_year=2025,
    )
)
````
