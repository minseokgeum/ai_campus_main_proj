```python
#원주가, 수정 주가 둘 다 한번에 조회
import os
import requests
import pandas as pd


API_TOKEN = os.getenv('TIINGO_API_TOKEN')

ticker = "AAPL"
start_date = "2025-10-01"
end_date = "2026-10-10"

url = f"https://api.tiingo.com/tiingo/daily/{ticker}/prices"

params = {
    "startDate": start_date,
    "endDate": end_date,
    "token": API_TOKEN,
}

response = requests.get(url, params=params)
response.raise_for_status()

data = response.json()

df = pd.DataFrame(data)

df
```

```python
#원주가 조회
raw_df = df[
    [
        "date",
        "open",
        "high",
        "low",
        "close",
        "volume",
    ]
].copy()

raw_df["date"] = pd.to_datetime(raw_df["date"]).dt.date

for col in ["open", "high", "low", "close"]:
    raw_df[col] = pd.to_numeric(raw_df[col])

raw_df["volume"] = pd.to_numeric(raw_df["volume"])

raw_df
```

```python
#수정주가
adj_df = df[
    [
        "date",
        "adjOpen",
        "adjHigh",
        "adjLow",
        "adjClose",
        "adjVolume",
    ]
].copy()

adj_df["date"] = pd.to_datetime(adj_df["date"]).dt.date

for col in ["adjOpen", "adjHigh", "adjLow", "adjClose"]:
    adj_df[col] = pd.to_numeric(adj_df[col])

adj_df["adjVolume"] = pd.to_numeric(adj_df["adjVolume"])

adj_df
```
