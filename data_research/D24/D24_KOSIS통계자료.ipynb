```python
import requests
import pandas as pd
from getpass import getpass

KOSIS_KEY = getpass("KOSIS API 인증키 입력: ")

SEARCH_URL = "https://kosis.kr/openapi/statisticsSearch.do"

params = {
    "method": "getList",
    "apiKey": KOSIS_KEY,
    "format": "json",
    "jsonVD": "Y",             # ← 이거 추가
    "searchNm": "산업생산지수",
    "sort": "RANK",
    "startCount": 1,
    "resultCount": 20
}

response = requests.get(
    SEARCH_URL,
    params=params,
    timeout=60
)

print("HTTP 상태코드:", response.status_code)
print("응답 앞부분:")
print(response.text[:500])
```

```python
data = response.json()

search_df = pd.DataFrame(data)

display(search_df)
```

```python
cols = [
    "ORG_ID",
    "ORG_NM",
    "TBL_ID",
    "TBL_NM",
    "STRT_PRD_DE",
    "END_PRD_DE"
]

display(
    search_df[
        [c for c in cols if c in search_df.columns]
    ]
)
```
